```python
from __future__ import annotations
from dataclasses import dataclass, asdict
from typing import Any, Dict, List
import json, math, random
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=6, suppress=True)
SEED = 20260322
random.seed(SEED)
np.random.seed(SEED)

PAPER_ID = "CF000"
PAPER_SOURCE_MODE = "manifest_only"
IO_MODE = "forbidden"
FIGURE_MODE = "inline_only"

@dataclass
class GateResult:
    gate_id: str
    claim_ids: List[str]
    section_id: str
    burden_type: str
    criteria: Dict[str, Any]
    metrics: Dict[str, Any]
    passed: bool
    figure_refs: List[str]
    figure_description: str
    notes: str = ""

GATE_LOG: List[GateResult] = []

def _json_default(o):
    if isinstance(o, np.integer):
        return int(o)
    if isinstance(o, np.floating):
        return float(o)
    if isinstance(o, np.bool_):
        return bool(o)
    return str(o)

def emit_gate(result: GateResult):
    GATE_LOG.append(result)
    payload = asdict(result)
    print(json.dumps(payload, indent=2, default=_json_default))
    return payload

print({
    "paper_id": PAPER_ID,
    "paper_source_mode": PAPER_SOURCE_MODE,
    "io_mode": IO_MODE,
    "figure_mode": FIGURE_MODE,
    "seed": SEED,
})
```

```python
class CF000WitnessModel:
    def __init__(self, axis1_depth_max: int = 9):
        self.axis1_depth_max = axis1_depth_max
        self.origin = "ω"
        self.candidates = ["0", "1", "ω"]
        self.stage_to_dim = {"0D":0, "1D":1, "2D":2}

    def realized(self, x: str) -> bool:
        return x == self.origin

    def non_vacuous(self, x: str) -> bool:
        return x != "0"

    def flat(self, x: str) -> bool:
        return x in {"0", "1"}

    def invariant_bearing(self, x: str) -> bool:
        return self.realized(x)

    def differentiated(self, x: str) -> bool:
        return self.realized(x) and (not self.flat(x))

    def degmin(self, x: str) -> bool:
        return x == self.origin and self.differentiated(x) and self.invariant_bearing(x)

    def same_axis_successor_count(self, depth: int) -> int:
        return 2 if depth < self.axis1_depth_max else 0

    def same_axis_states(self, max_depth: int = None):
        m = self.axis1_depth_max if max_depth is None else max_depth
        return [{"depth":d, "successors":self.same_axis_successor_count(d)} for d in range(m+1)]

    def invariant_stack(self, stage: str):
        if stage == "0D":
            return np.array([1,0,0])  # primitive, 1D, 2D effective invariants
        if stage == "1D":
            return np.array([1,1,0])
        if stage == "2D":
            return np.array([1,1,1])
        raise ValueError(stage)

MODEL = CF000WitnessModel(axis1_depth_max=9)
print({
    "origin": MODEL.origin,
    "axis1_depth_max": MODEL.axis1_depth_max,
    "stages": list(MODEL.stage_to_dim.keys()),
})
```

## Gate 12 — ORS, Pre-Metric Orthogonal Re-Articulation

**Target claim:** `4.6.1`

**Pass criteria**
- If the current axis is saturated, a new admissible same-level determination cannot be realized as more of that same axis.
- The new admissible determination must therefore occupy an irreducible new axis/domain.
- The gate passes only if the second direction is irreducible to the first.

**Figure requirement**
- Render the saturated first axis and the irreducible second axis.

```python
axis1 = np.array([1.0, 0.0])
axis2 = np.array([0.0, 1.0])
dot = float(np.dot(axis1, axis2))
irreducible_new_axis = abs(dot) < 1e-12

plt.figure(figsize=(4.5, 4.5))
plt.arrow(0, 0, axis1[0], axis1[1], head_width=0.05, length_includes_head=True)
plt.arrow(0, 0, axis2[0], axis2[1], head_width=0.05, length_includes_head=True)
plt.text(1.02, -0.02, "A1 saturated")
plt.text(-0.02, 1.02, "A2 irreducible")
plt.xlim(-0.2, 1.2)
plt.ylim(-0.2, 1.2)
plt.gca().set_aspect('equal', adjustable='box')
plt.title("Pre-metric orthogonal re-articulation")
plt.show()

pass_cond = irreducible_new_axis
emit_gate(GateResult(
    gate_id="G12",
    claim_ids=["4.6.1"],
    section_id="4.6",
    burden_type="orthogonal_rearticulation",
    criteria={
        "saturated_axis_cannot_host_new_same_axis_determination": True,
        "new_determination_requires_irreducible_axis": True
    },
    metrics={"axis_dot_product": dot, "irreducible_new_axis": irreducible_new_axis},
    passed=pass_cond,
    figure_refs=["inline://G12/ors_axes"],
    figure_description="Two-axis inline figure showing irreducible re-articulation from a saturated first axis.",
    notes="The figure is used only as a witness of irreducibility, not as imported Euclidean structure."
))
```

## Gate 13 — Same-Axis Bearing and Saturation

**Target claims:** `4.7.2`, `4.7.3`

**Pass criteria**
- While same-axis successors remain available, the axis is not saturated.
- At the saturation depth, successor count drops to zero.
- After saturation, continued genuinely new same-axis articulation is impossible.
- Because discharge is excluded, continued articulation must move orthogonally.

**Figure requirement**
- Render successor availability over 1D depth, with the saturation frontier marked.

```python
states = MODEL.same_axis_states()
depths = [row["depth"] for row in states]
succs = [row["successors"] for row in states]
sat_depths = [d for d, s in zip(depths, succs) if s == 0]
sat_depth = sat_depths[0]

plt.figure(figsize=(7.2, 3.6))
plt.step(depths, succs, where="mid")
plt.axvline(sat_depth, linestyle="--")
plt.text(sat_depth + 0.1, max(succs)*0.8, f"sat @ depth {sat_depth}")
plt.xlabel("1D articulation depth")
plt.ylabel("Same-axis successor count")
plt.title("Same-axis availability collapses at the saturation frontier")
plt.show()

pass_cond = (
    all(s > 0 for d, s in zip(depths, succs) if d < sat_depth) and
    succs[-1] == 0 and
    sat_depth == MODEL.axis1_depth_max
)
emit_gate(GateResult(
    gate_id="G13",
    claim_ids=["4.7.2", "4.7.3"],
    section_id="4.7",
    burden_type="saturation_forcing",
    criteria={
        "pre_saturation_same_axis_successors_positive": True,
        "at_saturation_successors_zero": True,
        "post_saturation_new_same_axis_articulation_impossible": True,
        "continued_articulation_must_move_orthogonally": True
    },
    metrics={
        "axis1_depth_max": MODEL.axis1_depth_max,
        "saturation_depth": sat_depth,
        "successor_sequence": succs
    },
    passed=pass_cond,
    figure_refs=["inline://G13/saturation_frontier"],
    figure_description="Step plot of same-axis successor availability with the saturation frontier marked.",
    notes="This gate is the discrete witness for same-axis exhaustion without discharge."
))
```
