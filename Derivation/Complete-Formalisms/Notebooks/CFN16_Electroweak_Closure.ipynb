```python

from __future__ import annotations

from dataclasses import dataclass
from typing import Any, Dict, List
import hashlib
import math
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=10, suppress=True)

CF16_PATH = Path(r"/mnt/data/CF16.tex")
A_MINUS_ONE_PATH = Path(r"/mnt/data/A(-1).tex")
TEMPLATE_PATH = Path(r"/mnt/data/CFNXX_strict-numerical-gates-template.ipynb")

@dataclass
class GateResult:
    gate_id: str
    gate_name: str
    passed: bool
    metrics: Dict[str, Any]
    pass_criteria: Dict[str, Any]
    notes: str

LEDGER: List[GateResult] = []

def terminal_log(gate_id: str, title: str, passed: bool, metrics: Dict[str, Any], criteria: Dict[str, Any], notes: str = "") -> None:
    status_symbol = "✅" if passed else "❌"
    print("=" * 88)
    print(f"[{gate_id}] {title}")
    print("- status:", f"{status_symbol} {'PASS' if passed else 'FAIL'}")
    print("- criteria:")
    for k, v in criteria.items():
        print(f"    * {k}: {v}")
    print("- metrics:")
    for k, v in metrics.items():
        print(f"    * {k}: {v}")
    if notes:
        print("- notes:", notes)
    print("=" * 88)


def add_gate(gate_id: str, gate_name: str, passed: bool, metrics: Dict[str, Any], criteria: Dict[str, Any], notes: str = "") -> None:
    terminal_log(gate_id, gate_name, passed, metrics, criteria, notes)
    LEDGER.append(GateResult(gate_id, gate_name, passed, metrics, criteria, notes))


def sha256_hex(path: Path) -> str:
    return hashlib.sha256(path.read_bytes()).hexdigest()


criteria = {
    "imports_available": True,
    "ledger_initialized": True,
    "inline_figure_rendered": True,
    "stdout_terminal_log_present": True,
    "filesystem_io_required": False,
}
metrics = {
    "imports_available": True,
    "ledger_length_after_init": len(LEDGER),
    "filesystem_io_required": False,
}

fig, ax = plt.subplots(figsize=(6.0, 3.2))
ax.bar(["imports", "ledger", "inline fig", "stdout", "no I/O"], [1, 1, 1, 1, 1])
ax.set_ylim(0, 1.2)
ax.set_title("Notebook substrate contract check")
ax.set_ylabel("satisfied = 1")
ax.grid(alpha=0.25)
plt.show()

passed = all([
    metrics["imports_available"],
    metrics["ledger_length_after_init"] == 0,
    not metrics["filesystem_io_required"],
])
add_gate(
    "T0",
    "Notebook substrate integrity",
    passed,
    metrics,
    criteria,
    notes="Helper layer, visible plot path, and ledger initialization all executed in the open.",
)
```

This cell does not reward the paper for merely naming $SU(2)$. It asks for operational evidence that a phase companion is algebraically distinct from the two-mode internal rotation, and that a purely abelian action is too weak to generate generic doublet structure.

**Pass criteria**
- phase × special-unitary factorization residual is below `1e-12`,
- constructed $SU(2)$ orbit reaches sampled normalized targets with error below `1e-12`,
- U(1)-only action misses mixed targets by a visible margin.

**Fail criteria**
- factorization fails,
- $SU(2)$ cannot reach generic targets,
- U(1)-only action reaches the same targets just as well.

**This cell cannot prove**
- abstract uniqueness among every possible group-theoretic construction. It attacks the concrete minimal two-mode hosting burden CF16 actually uses.

```python

rng = np.random.default_rng(20260322)


def random_unitary_2(rng: np.random.Generator) -> np.ndarray:
    z = rng.normal(size=(2, 2)) + 1j * rng.normal(size=(2, 2))
    q, r = np.linalg.qr(z)
    phases = np.exp(-1j * np.angle(np.diag(r)))
    return q @ np.diag(phases)


def random_state_2(rng: np.random.Generator) -> np.ndarray:
    z = rng.normal(size=2) + 1j * rng.normal(size=2)
    z /= np.linalg.norm(z)
    return z


n_u = 128
factorization_residuals = []
su2_det_residuals = []

for _ in range(n_u):
    U = random_unitary_2(rng)
    phi = 0.5 * np.angle(np.linalg.det(U))
    S = np.exp(-1j * phi) * U
    factorization_residuals.append(np.linalg.norm(U - np.exp(1j * phi) * S))
    su2_det_residuals.append(abs(np.linalg.det(S) - 1.0))

n_states = 256
target_beta_amp = []
su2_reach_errors = []
u1_only_errors = []
e1 = np.array([1.0 + 0.0j, 0.0 + 0.0j])

for _ in range(n_states):
    psi = random_state_2(rng)
    alpha, beta = psi[0], psi[1]
    Spsi = np.array([[alpha, -np.conj(beta)], [beta, np.conj(alpha)]], dtype=complex)
    su2_reach_errors.append(np.linalg.norm(Spsi @ e1 - psi))
    target_beta_amp.append(abs(beta))
    u1_only_errors.append(np.linalg.norm(np.exp(1j * np.angle(alpha)) * e1 - psi))

target_beta_amp = np.array(target_beta_amp)
su2_reach_errors = np.array(su2_reach_errors)
u1_only_errors = np.array(u1_only_errors)

mixed_mask = target_beta_amp > 0.20
metrics = {
    "su2_factorization_residual_max": float(np.max(factorization_residuals)),
    "su2_det_residual_max": float(np.max(su2_det_residuals)),
    "su2_orbit_reach_error_max": float(np.max(su2_reach_errors)),
    "u1_only_miss_mean": float(np.mean(u1_only_errors[mixed_mask])),
    "u1_only_miss_min_on_mixed_targets": float(np.min(u1_only_errors[mixed_mask])),
}
criteria = {
    "su2_factorization_residual_max": "<= 1e-12",
    "su2_det_residual_max": "<= 1e-12",
    "su2_orbit_reach_error_max": "<= 1e-12",
    "u1_only_miss_mean": ">= 0.2",
}
passed = (
    metrics["su2_factorization_residual_max"] <= 1e-12
    and metrics["su2_det_residual_max"] <= 1e-12
    and metrics["su2_orbit_reach_error_max"] <= 1e-12
    and metrics["u1_only_miss_mean"] >= 0.2
)

fig, ax = plt.subplots(figsize=(7.0, 3.6))
order = np.argsort(target_beta_amp)
ax.plot(target_beta_amp[order], su2_reach_errors[order], label="SU(2) reach error")
ax.plot(target_beta_amp[order], u1_only_errors[order], label="U(1)-only miss error")
ax.set_title("Two-mode target reach: SU(2) versus U(1)-only action")
ax.set_xlabel(r"target second-mode amplitude $|\beta|$")
ax.set_ylabel("state-space error")
ax.legend()
ax.grid(alpha=0.25)
plt.show()

add_gate(
    "E1",
    "Minimal electroweak-hosting fiber",
    passed,
    metrics,
    criteria,
    notes="The abelian companion separates as a phase burden, while SU(2) reaches generic normalized doublets and U(1)-only action does not.",
)
```

## Gate E2 — Chiral weak coupling from boundary hosting

**Plain-language view**

CF16 inherits its left-only weak action from CF08/CF13. The point is simple: a common weak action can only act on modes that actually share the active boundary. If both handed families co-live on the same wall, the chiral restriction is fake.

**Claim being audited:** the active chiral family stays strongly supported on the active wall while partner leakage into that same wall decays exponentially with wall separation.

**Canon anchor:** CF16 §5; Lemma `chiral_hosting`; Theorem `ew_closure`, clause 2; imported wall-hosting logic from CF08/CF13.

**Burden type:** localization / hosting asymmetry.

**Why this is an honest attack**

This cell attacks the exact structural hinge CF16 uses: shared weak action requires shared hosting. It does not assume chirality by label. It measures boundary support and partner leakage directly.

**Pass criteria**
- active-family support on the active wall stays above `0.95`,
- partner support on that wall falls below `1e-6` at large separation,
- partner leakage is exponentially fit with `R^2 >= 0.995`,
- coincident walls show no fake selectivity.

**Fail criteria**
- partner leakage does not collapse with separation,
- active-family support on the chosen wall is not robust,
- coincident walls already look selectively chiral.

**This cell cannot prove**
- the full overlap-operator or Ginsparg--Wilson construction. It attacks only the wall-hosting asymmetry CF16 relies on.

```python

def sech(x: np.ndarray) -> np.ndarray:
    return 1.0 / np.cosh(x)

xi = 1.0
profile_power = 2.0
z = np.linspace(-12.0, 12.0, 40001)
separations = np.linspace(0.0, 8.0, 17)
left_support = []
right_support = []

for L in separations:
    psi_left = sech((z + L / 2.0) / xi) ** profile_power
    psi_right = sech((z - L / 2.0) / xi) ** profile_power
    psi_left /= math.sqrt(np.trapezoid(np.abs(psi_left) ** 2, z))
    psi_right /= math.sqrt(np.trapezoid(np.abs(psi_right) ** 2, z))

    window_halfwidth = 1.5 * xi
    active_mask = (z >= (-L / 2.0 - window_halfwidth)) & (z <= (-L / 2.0 + window_halfwidth))
    left_support.append(float(np.trapezoid(np.abs(psi_left[active_mask]) ** 2, z[active_mask])))
    right_support.append(float(np.trapezoid(np.abs(psi_right[active_mask]) ** 2, z[active_mask])))

left_support = np.array(left_support)
right_support = np.array(right_support)

fit_mask = separations >= 2.0
coef = np.polyfit(separations[fit_mask], np.log(right_support[fit_mask]), 1)
pred = np.polyval(coef, separations[fit_mask])
ss_res = float(np.sum((np.log(right_support[fit_mask]) - pred) ** 2))
ss_tot = float(np.sum((np.log(right_support[fit_mask]) - np.mean(np.log(right_support[fit_mask]))) ** 2))
r2 = 1.0 - ss_res / ss_tot

metrics = {
    "left_support_min": float(np.min(left_support[1:])),
    "right_support_at_max_sep": float(right_support[-1]),
    "right_leakage_fit_r2": float(r2),
    "coincident_selectivity_gap": float(abs(left_support[0] - right_support[0])),
}
criteria = {
    "left_support_min": ">= 0.95",
    "right_support_at_max_sep": "<= 1e-6",
    "right_leakage_fit_r2": ">= 0.995",
    "coincident_selectivity_gap": "<= 1e-12",
}
passed = (
    metrics["left_support_min"] >= 0.95
    and metrics["right_support_at_max_sep"] <= 1e-6
    and metrics["right_leakage_fit_r2"] >= 0.995
    and metrics["coincident_selectivity_gap"] <= 1e-12
)

fig, ax = plt.subplots(figsize=(7.0, 3.8))
ax.semilogy(separations, left_support, marker="o", label="active-family support on active wall")
ax.semilogy(separations, right_support, marker="o", label="partner leakage onto active wall")
ax.set_title("Boundary-hosting selectivity versus wall separation")
ax.set_xlabel("wall separation L / xi")
ax.set_ylabel("support fraction inside active-wall window")
ax.legend()
ax.grid(alpha=0.25, which="both")
plt.show()

add_gate(
    "E2",
    "Chiral weak coupling from boundary hosting",
    passed,
    metrics,
    criteria,
    notes=f"log-leakage slope={coef[0]:.6f}; large-separation partner leakage collapses while coincident walls remain symmetric.",
)
```

## Gate E3 — Neutral mixing into one massless and one massive direction

**Plain-language view**

The neutral sector is where the electroweak architecture either locks or fails. CF16 says the broken neutral matrix is rank one: one exactly massless direction survives, one orthogonal direction becomes massive, and the mixing angle is the diagonalizing rotation.

**Claim being audited:** the claimed neutral mass matrix has one null mode, one heavy mode, and the advertised mixing directions, while a nearby corrupted control loses the zero mode.

**Canon anchor:** CF16 §5; Theorem `ew_closure`, clauses 4--5.

**Burden type:** exact eigensystem / structural diagonalization.

**Why this is an honest attack**

This is not a soft plausibility check. It measures the exact null-mode and heavy-mode burdens across a parameter sweep and asks whether a nearby broken matrix would still pass. It should not.

**Pass criteria**
- the smallest true eigenvalue stays below `1e-12`,
- the heavy-mode residual stays below `1e-12`,
- the corrupted control has smallest eigenvalue above `1e-3`.

**Fail criteria**
- the true matrix is not exactly rank one,
- the claimed heavy mode is wrong,
- the corrupted matrix also keeps an exact null mode.

**This cell cannot prove**
- loop corrections or phenomenology beyond the structural/tree-level neutral-mixing layer.

```python

a_vals = np.linspace(0.3, 2.0, 12)
b_vals = np.linspace(0.2, 1.8, 11)
nu_vals = np.linspace(0.5, 2.0, 10)

true_smallest = []
massive_residuals = []
corrupted_smallest = []

for a in a_vals:
    for b in b_vals:
        for nu in nu_vals:
            M = (nu ** 2) * np.array([[a ** 2, -a * b], [-a * b, b ** 2]], dtype=float)
            evals = np.linalg.eigvalsh(M)
            true_smallest.append(abs(evals[0]))

            lam_heavy = nu ** 2 * (a ** 2 + b ** 2)
            Z = np.array([a, -b], dtype=float) / math.sqrt(a ** 2 + b ** 2)
            massive_residuals.append(np.linalg.norm(M @ Z - lam_heavy * Z))

            M_bad = (nu ** 2) * np.array([[a ** 2, -0.8 * a * b], [-0.8 * a * b, b ** 2]], dtype=float)
            corrupted_smallest.append(float(np.min(np.linalg.eigvalsh(M_bad))))

true_smallest = np.array(true_smallest)
massive_residuals = np.array(massive_residuals)
corrupted_smallest = np.array(corrupted_smallest)

metrics = {
    "true_smallest_eigenvalue_max": float(np.max(true_smallest)),
    "massive_mode_residual_max": float(np.max(massive_residuals)),
    "corrupted_smallest_eigenvalue_min": float(np.min(corrupted_smallest)),
}
criteria = {
    "true_smallest_eigenvalue_max": "<= 1e-12",
    "massive_mode_residual_max": "<= 1e-12",
    "corrupted_smallest_eigenvalue_min": ">= 1e-3",
}
passed = (
    metrics["true_smallest_eigenvalue_max"] <= 1e-12
    and metrics["massive_mode_residual_max"] <= 1e-12
    and metrics["corrupted_smallest_eigenvalue_min"] >= 1e-3
)

fig, ax = plt.subplots(figsize=(7.0, 3.6))
ax.plot(np.sort(true_smallest), label="true smallest eigenvalue")
ax.plot(np.sort(corrupted_smallest), label="corrupted smallest eigenvalue")
ax.set_yscale("log")
ax.set_title("Neutral-sector null mode versus corrupted control")
ax.set_xlabel("sweep sample (sorted)")
ax.set_ylabel("smallest eigenvalue")
ax.legend()
ax.grid(alpha=0.25, which="both")
plt.show()

add_gate(
    "E3",
    "Neutral mixing into one massless and one massive direction",
    passed,
    metrics,
    criteria,
    notes="The advertised rank-one neutral matrix keeps one exact null direction; the corrupted off-diagonal coupling does not.",
)
```

## Gate E4 — Charged/neutral mass law from one shared order parameter

**Plain-language view**

CF16 says the charged and neutral massive sectors are not unrelated numbers. They come from the same broken order parameter, and that forces the tree-level structural relation $m_W = m_Z \cos	heta_{\mathrm{EW}}$.

**Claim being audited:** the advertised $W/Z$ mass law holds exactly across a shared-order-parameter sweep, while a corrupted neutral-mass law fails visibly.

**Canon anchor:** CF16 §5; Theorem `ew_closure`, clause 6; equation `wz_relation`.

**Burden type:** exact mass relation / shared-order-parameter scaling.

**Why this is an honest attack**

This cell does not merely restate the formula. It checks exact equality across a nontrivial $
u$ sweep and separately attacks the common-origin claim by verifying linear scaling with the same order amplitude.

**Pass criteria**
- maximum $m_W - m_Z \cos	heta$ residual is below `1e-12`,
- log-slope error of $m_W$ versus $
u$ is below `1e-12`,
- log-slope error of $m_Z$ versus $
u$ is below `1e-12`,

**Fail criteria**
- the equality is not exact,

**This cell cannot prove**

```python

a = 1.3
b = 0.7
nu = np.linspace(0.2, 3.0, 60)
cos_theta = a / math.sqrt(a ** 2 + b ** 2)

mW = nu * a
mZ = nu * math.sqrt(a ** 2 + b ** 2)
mZ_proj = mZ * cos_theta
mZ_bad = nu * (a + b)

mass_relation_residual = np.max(np.abs(mW - mZ_proj))
mw_log_slope_error = abs(np.polyfit(np.log(nu), np.log(mW), 1)[0] - 1.0)
mz_log_slope_error = abs(np.polyfit(np.log(nu), np.log(mZ), 1)[0] - 1.0)
corrupted_relation_rms = float(np.sqrt(np.mean((mW - mZ_bad * cos_theta) ** 2)))

metrics = {
    "mass_relation_residual_max": float(mass_relation_residual),
    "mw_log_slope_error": float(mw_log_slope_error),
    "mz_log_slope_error": float(mz_log_slope_error),
    "corrupted_relation_rms": corrupted_relation_rms,
}
criteria = {
    "mass_relation_residual_max": "<= 1e-12",
    "mw_log_slope_error": "<= 1e-12",
    "mz_log_slope_error": "<= 1e-12",
    "corrupted_relation_rms": ">= 0.1",
}
passed = (
    metrics["mass_relation_residual_max"] <= 1e-12
    and metrics["mw_log_slope_error"] <= 1e-12
    and metrics["mz_log_slope_error"] <= 1e-12
    and metrics["corrupted_relation_rms"] >= 0.1
)

fig, ax = plt.subplots(figsize=(7.0, 3.8))
ax.plot(nu, mW, label=r"$m_W$")
ax.plot(nu, mZ_proj, label=r"$m_Z \cos 	heta_{EW}$")
ax.plot(nu, mZ_bad * cos_theta, linestyle="--", label="corrupted comparison")
ax.set_title("Structural W/Z mass law over the shared order amplitude")
ax.set_xlabel(r"order amplitude $\nu$")
ax.set_ylabel("mass scale")
ax.legend()
ax.grid(alpha=0.25)
plt.show()

add_gate(
    "E4",
    "Charged/neutral mass law from one shared order parameter",
    passed,
    metrics,
    criteria,
    notes="Both masses scale linearly with the same order amplitude, and the structural W/Z law holds exactly at the advertised layer.",
)
```

## Gate E5 — Residual electromagnetic charge law

**Plain-language view**

The surviving massless neutral direction has to leave the condensed lower component invisible. That is what fixes the charge law. If nearby coefficients also keep the condensate neutral, the paper's uniqueness claim is not real.

**Claim being audited:** condensate neutrality fixes the residual generator to $Q_{\mathrm{em}} = T_3 + Y/2$ after companion normalization.

**Canon anchor:** CF16 §3; Definition `em_charge`; Lemma `qem_unique`; Theorem `ew_closure`, clause 3.

**Burden type:** exact neutrality constraint / coefficient fixing.

**Why this is an honest attack**

This cell starts from the most general residual generator $Q = T_3 + cY$ and solves the neutrality burden instead of assuming the answer in advance. It then attacks uniqueness with nearby wrong coefficients.

**Pass criteria**
- condensate neutrality residual is below `1e-12`,
- the upper doublet component has charge `+1`,
- the lower doublet component has charge `0`,
- nearby wrong coefficients leave at least `0.05` charge magnitude on the condensate.

**Fail criteria**
- neutrality does not isolate the coefficient,
- the normalized law does not give the advertised charges,
- wrong coefficients also keep the condensate neutral.

**This cell cannot prove**
- every downstream matter-sector charge assignment. It attacks the minimal weak-doublet residual charge law CF16 claims.

```python

T3 = np.diag([0.5, -0.5])
e_plus = np.array([1.0, 0.0])
e_zero = np.array([0.0, 1.0])

y_phi_values = np.linspace(0.25, 2.0, 32)
neutrality_residuals = []

for y_phi in y_phi_values:
    Y = y_phi * np.eye(2)
    c = 1.0 / (2.0 * y_phi)
    Q = T3 + c * Y
    neutrality_residuals.append(np.linalg.norm(Q @ e_zero))

Y_norm = np.eye(2)
Q_em = T3 + 0.5 * Y_norm
upper_component_charge = float(np.real(e_plus @ (Q_em @ e_plus)))
lower_component_charge = float(np.real(e_zero @ (Q_em @ e_zero)))

c_grid = np.linspace(0.0, 1.0, 401)
wrong_coeff_charges = np.array([np.linalg.norm((T3 + c * Y_norm) @ e_zero) for c in c_grid])
wrong_mask = np.abs(c_grid - 0.5) >= 0.05

metrics = {
    "neutrality_residual_max": float(np.max(neutrality_residuals)),
    "upper_component_charge": upper_component_charge,
    "lower_component_charge": lower_component_charge,
    "wrong_coefficient_charge_min": float(np.min(wrong_coeff_charges[wrong_mask])),
}
criteria = {
    "neutrality_residual_max": "<= 1e-12",
    "upper_component_charge": "== 1",
    "lower_component_charge": "== 0",
    "wrong_coefficient_charge_min": ">= 0.05",
}
passed = (
    metrics["neutrality_residual_max"] <= 1e-12
    and abs(metrics["upper_component_charge"] - 1.0) <= 1e-12
    and abs(metrics["lower_component_charge"] - 0.0) <= 1e-12
    and metrics["wrong_coefficient_charge_min"] >= 0.05
)

fig, ax = plt.subplots(figsize=(7.0, 3.6))
ax.plot(c_grid, wrong_coeff_charges)
ax.axvline(0.5, linestyle="--", label=r"$c = 1/2$")
ax.set_title("Condensate charge magnitude versus residual-generator coefficient")
ax.set_xlabel(r"coefficient $c$ in $Q = T_3 + cY$")
ax.set_ylabel(r"$\|Q \phi_0\|$")
ax.legend()
ax.grid(alpha=0.25)
plt.show()

add_gate(
    "E5",
    "Residual electromagnetic charge law",
    passed,
    metrics,
    criteria,
    notes="Neutrality isolates the coefficient uniquely, and the normalized doublet charges become (+1, 0) as advertised.",
)
```

## Gate E6 — Only the normal electroweak interface mode remains as a physical scalar

**Plain-language view**

The broken electroweak layer starts with a four-real-component order parameter. CF16 says the minimal isotropic broken sector leaves three tangential directions as flat gauge-carried modes and only one normal radial mode as a physical scalar.

**Claim being audited:** the vacuum Hessian of the minimal isotropic order potential has exactly three zero tangential modes and one positive radial mode aligned with the vacuum direction, while an anisotropic control leaves an extra scalar.

**Canon anchor:** CF16 §5; Theorem `ew_closure`, clause 7.

**Burden type:** exact mode counting / scalar closure.

**Why this is an honest attack**

This cell does not name a Higgs field first and then check a box. It computes the vacuum mode spectrum directly and attacks the uniqueness claim with a control that deliberately leaves an extra scalar.

**Pass criteria**
- the isotropic vacuum has exactly three zero modes,
- it has exactly one positive mode,
- the positive-mode eigenvector aligns with the vacuum direction with cosine at least `0.999999`,
- the anisotropic control creates at least one extra positive scalar mode.

**Fail criteria**
- more than one positive scalar survives in the minimal isotropic layer,
- the heavy mode is not the normal direction,
- the control does not separate from the minimal layer.

**This cell cannot prove**
- full gauge fixing or quantum corrections. It attacks the local broken-layer mode-counting burden only.

```python

lam = 0.8
v = 2.0
phi_vac = np.array([0.0, 0.0, 0.0, v / math.sqrt(2.0)])


def higgs_hessian(phi: np.ndarray, lam: float, v: float) -> np.ndarray:
    x = float(phi @ phi)
    return 4.0 * lam * ((x - v * v / 2.0) * np.eye(4) + 2.0 * np.outer(phi, phi))


H_true = higgs_hessian(phi_vac, lam=lam, v=v)
true_evals, true_evecs = np.linalg.eigh(H_true)

# anisotropic control that leaves one extra scalar in the first tangential direction
anisotropy = 1.5
H_ctrl = H_true.copy()
H_ctrl[0, 0] += 2.0 * anisotropy
ctrl_evals = np.linalg.eigvalsh(H_ctrl)

zero_tol = 1e-10
zero_mode_count = int(np.sum(np.abs(true_evals) <= zero_tol))
positive_mode_count = int(np.sum(true_evals > zero_tol))
control_extra_scalar_count = int(np.sum(ctrl_evals > zero_tol) - positive_mode_count)
radial_vec = phi_vac / np.linalg.norm(phi_vac)
heavy_vec = true_evecs[:, np.argmax(true_evals)]
radial_alignment = float(abs(np.vdot(radial_vec, heavy_vec)))

metrics = {
    "zero_mode_count": zero_mode_count,
    "positive_mode_count": positive_mode_count,
    "radial_alignment": radial_alignment,
    "control_extra_scalar_count": control_extra_scalar_count,
}
criteria = {
    "zero_mode_count": 3,
    "positive_mode_count": 1,
    "radial_alignment": ">= 0.999999",
    "control_extra_scalar_count": ">= 1",
}
passed = (
    metrics["zero_mode_count"] == 3
    and metrics["positive_mode_count"] == 1
    and metrics["radial_alignment"] >= 0.999999
    and metrics["control_extra_scalar_count"] >= 1
)

fig, ax = plt.subplots(figsize=(7.0, 3.6))
ax.plot(np.sort(true_evals), marker="o", label="minimal isotropic layer")
ax.plot(np.sort(ctrl_evals), marker="o", label="anisotropic control")
ax.axhline(0.0, color="black", linewidth=1.0)
ax.set_title("Broken-layer scalar spectrum: minimal vacuum versus control")
ax.set_xlabel("sorted mode index")
ax.set_ylabel("Hessian eigenvalue")
ax.legend()
ax.grid(alpha=0.25)
plt.show()

add_gate(
    "E6",
    "Only the normal electroweak interface mode remains as a physical scalar",
    passed,
    metrics,
    criteria,
    notes="The minimal isotropic layer yields three tangential zero modes and one radial heavy mode; the anisotropic control leaves an extra scalar.",
)
```

```python
gate_ids = [g.gate_id for g in LEDGER]
pass_count = sum(int(g.passed) for g in LEDGER)
fail_count = len(LEDGER) - pass_count
covered_paper_gates = {"G1": "E1", "G2": "E2", "G3": "E3", "G4": "E4", "G5": "E5", "G6": "E6"}
missing_paper_gates = [g for g in covered_paper_gates if covered_paper_gates[g] not in gate_ids]

fig, ax = plt.subplots(figsize=(8.0, 3.6))
ax.bar(gate_ids, [1 if g.passed else 0 for g in LEDGER])
ax.set_ylim(0, 1.2)
ax.set_title("CF16 final results ledger")
ax.set_ylabel("PASS = 1, FAIL = 0")
ax.grid(alpha=0.25)
plt.show()

metrics = {
    "ledger_gate_count": len(LEDGER),
    "unique_gate_count": len(set(gate_ids)),
    "pass_count": pass_count,
    "fail_count": fail_count,
    "missing_paper_gate_count": len(missing_paper_gates),
}
criteria = {
    "ledger_gate_count": ">= 1",
    "unique_gate_count_equals_ledger_gate_count": True,
    "pass_count_plus_fail_count_equals_ledger_gate_count": True,
    "missing_paper_gate_count": 0,
}
passed = (
    len(LEDGER) >= 1
    and len(set(gate_ids)) == len(LEDGER)
    and (pass_count + fail_count == len(LEDGER))
    and len(missing_paper_gates) == 0
)
add_gate(
    "T6",
    "Notebook-wide final results ledger",
    passed,
    metrics,
    criteria,
    notes=f"paper-level coverage map={covered_paper_gates}; missing_paper_gates={missing_paper_gates}",
)

print("\nFINAL LEDGER TABLE")
print("-" * 88)
for g in LEDGER:
    print(f"{g.gate_id:>3} | {'PASS' if g.passed else 'FAIL':<4} | {g.gate_name}")
print("-" * 88)
print(f"TOTAL: {len(LEDGER)} gates | PASS={sum(int(g.passed) for g in LEDGER)} | FAIL={len(LEDGER)-sum(int(g.passed) for g in LEDGER)}")
print("\nCF16 PAPER-LEVEL GATE COVERAGE")
for paper_gate, notebook_gate in covered_paper_gates.items():
    print(f"- {paper_gate} -> {notebook_gate} ({'present' if notebook_gate in gate_ids else 'missing'})")
```
