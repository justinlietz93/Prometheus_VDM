# CF14 — Validation Audit Notebook

This notebook instantiates the strict numerical-gates template for **CF14: Principle of Stationary Action as Temporal Effective Invariant**.

The notebook is built as an auditor, not as a paper clone. It attacks theorem-bearing and descendant claims with executable residuals, negative controls, and thresholded gates. It does **not** embed bulk TeX or long source dumps.

## Audit model for this instantiated notebook

The notebook follows the paper in burden order.

- First it proves the notebook substrate is honest.
- Then it declares the exact CF14 burden contract in a compact manifest.
- Then it attacks executable claims: primitive stationarity, Euler–Lagrange balance, thermodynamic Legendre stationarity, Onsager's dissipative variational law, generic field stationarity, Maxwell stationarity, stationary phase, and Rayleigh–Ritz.
- It ends with a visible ledger that names what was covered and what CF14-advertised gates remain uncovered.

Hard anti-bloat rule enforced here: the manifest carries only claim-level burdens and source fingerprints. The paper body stays outside the notebook.

## Gate T0 — Notebook substrate integrity

**Plain-language view**

If the helper layer is weak, everything after it is contaminated. This cell checks that the instantiated notebook starts cleanly: imports work, the ledger exists, figures render inline, and the notebook can report hard pass/fail outcomes without filesystem tricks.

**Claim being audited:** the notebook substrate is strong enough to host a real CF14 audit.

**Why this is an honest attack**

This directly attacks the failure mode where a notebook hides behind “setup” cells that do not behave like audit cells. If the substrate cannot emit figures, logs, and structured results, the notebook is not ready to audit CF14.

**Pass criteria**
- helper imports succeed,
- the result ledger is initialized,
- an inline figure is rendered,
- terminal-style PASS / FAIL logs print,
- no file output is required for the audit logic itself.

**Fail criteria**
- missing helper imports,
- missing ledger,
- no inline figure,
- no terminal log,
- substrate depends on external writes.

**This cell cannot prove**
- that later CF14-specific attacks are strong enough. It only proves the notebook starts from an honest substrate.

```python
from __future__ import annotations

from dataclasses import dataclass, asdict
from typing import Any, Dict, List, Sequence
from pathlib import Path
import hashlib
import math

import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=10, suppress=True)

@dataclass
class GateResult:
    gate_id: str
    gate_name: str
    canon_anchor: str
    paper_gate_ids: List[str]
    passed: bool
    metrics: Dict[str, Any]
    pass_criteria: Dict[str, Any]
    notes: str
    partial: bool = False

LEDGER: List[GateResult] = []


def terminal_log(gate_id: str, title: str, passed: bool, metrics: Dict[str, Any], criteria: Dict[str, Any], notes: str = "") -> None:
    status_symbol = "✅" if passed else "❌"
    print("=" * 88)
    print(f"[{gate_id}] {title}")
    print("- status:", f"{status_symbol} {'PASS' if passed else 'FAIL'}")
    print("- criteria:")
    for k, v in criteria.items():
        print(f"    * {k}: {v}")
    print("- metrics:")
    for k, v in metrics.items():
        print(f"    * {k}: {v}")
    if notes:
        print("- notes:", notes)
    print("=" * 88)


def append_gate(gate_id: str, gate_name: str, canon_anchor: str, paper_gate_ids: Sequence[str], passed: bool, metrics: Dict[str, Any], pass_criteria: Dict[str, Any], notes: str = "", partial: bool = False) -> None:
    LEDGER.append(GateResult(
        gate_id=gate_id,
        gate_name=gate_name,
        canon_anchor=canon_anchor,
        paper_gate_ids=list(paper_gate_ids),
        passed=bool(passed),
        metrics=metrics,
        pass_criteria=pass_criteria,
        notes=notes,
        partial=bool(partial),
    ))


def sha256_of(path: str) -> str:
    return hashlib.sha256(Path(path).read_bytes()).hexdigest()


fig, ax = plt.subplots(figsize=(5.6, 2.6))
probe_x = np.linspace(0.0, 2.0 * np.pi, 400)
probe_y = np.sin(probe_x)
ax.plot(probe_x, probe_y)
ax.set_title("T0 substrate probe figure")
ax.set_xlabel("x")
ax.set_ylabel("sin(x)")
ax.grid(alpha=0.25)
plt.show()

criteria = {
    "imports_available": True,
    "ledger_initialized": True,
    "inline_figure_rendered": True,
    "stdout_terminal_log_present": True,
    "filesystem_io_required": False,
}
metrics = {
    "imports_available": True,
    "ledger_length_at_start": len(LEDGER),
    "probe_sample_count": int(probe_x.size),
    "filesystem_io_required": False,
}
passed = True
terminal_log("T0", "Notebook substrate integrity", passed, metrics, criteria, notes="CF14 audit helper layer initialized.")
append_gate("T0", "Notebook substrate integrity", "notebook helper substrate", [], passed, metrics, criteria, notes="CF14 audit helper layer initialized.")
```

## Gate CF14-G3-1 — Primitive stationarity on an endpoint-fixed mechanical descendant

**Plain-language view**

This is the first executable core of CF14. If the stationarity theorem is real, an exact solution path should have vanishing first variation, while a nearby endpoint-fixed corrupted path should not.

**Canon anchor**
- CF14 §4, Primitive stationarity theorem
- `t h m:primitive_stationarity`
- `eq:primitive_stationarity`

**Burden type**
- theorem-bearing variational stationarity
- conservative realized descendant

**Exact statement being attacked**
For endpoint-fixed admissible variations, selected paths satisfy `δS_Inv[γ] = 0`.

**Why this attack is honest**
The notebook uses the realized mechanical descendant `L = T - V = 1/2 (qdot^2 - q^2)` on `[0, π]` with fixed endpoints `q(0)=q(π)=0`. The exact solution `q*(t)=sin(t)` should be stationary. A nearby endpoint-fixed corrupted path should keep a finite first-variation residual as the grid is refined.

**Pass criteria**
- the exact path's first-variation residual decreases with resolution,
- the finest-grid exact-path residual is below threshold,
- the corrupted path stays visibly nonzero on the finest grid.

**Fail criteria**
- the exact path does not approach stationarity,
- the corrupted path is not separated,
- both curves behave the same.

**This cell cannot prove**
- the full constitutional bridge to A(-1). It audits the theorem on a clean realized descendant.

```python
resolutions = np.array([101, 201, 401, 801, 1601], dtype=int)
mode_ids = [1, 2, 3, 4, 5]
h = 1e-4


def mechanical_action(q: np.ndarray, t: np.ndarray) -> float:
    qdot = np.gradient(q, t, edge_order=2)
    return float(np.trapezoid(0.5 * (qdot**2 - q**2), t))


def variation_probe(t: np.ndarray, mode_id: int) -> np.ndarray:
    return (t * (np.pi - t) / np.pi**2) * np.sin(mode_id * t)

true_residuals = []
wrong_residuals = []

for N in resolutions:
    t = np.linspace(0.0, np.pi, N)
    q_true = np.sin(t)
    q_wrong = np.sin(t) + 0.15 * (t * (np.pi - t) / np.pi**2)
    mode_res_true = []
    mode_res_wrong = []
    for n in mode_ids:
        eta = variation_probe(t, n)
        d_true = (mechanical_action(q_true + h * eta, t) - mechanical_action(q_true - h * eta, t)) / (2.0 * h)
        d_wrong = (mechanical_action(q_wrong + h * eta, t) - mechanical_action(q_wrong - h * eta, t)) / (2.0 * h)
        mode_res_true.append(abs(d_true))
        mode_res_wrong.append(abs(d_wrong))
    true_residuals.append(max(mode_res_true))
    wrong_residuals.append(max(mode_res_wrong))

true_residuals = np.array(true_residuals)
wrong_residuals = np.array(wrong_residuals)

fig, ax = plt.subplots(figsize=(6.8, 3.4))
ax.loglog(resolutions, true_residuals, marker="o", label="exact path")
ax.loglog(resolutions, wrong_residuals, marker="o", label="corrupted path")
ax.set_title("CF14-G3 witness: first-variation residual vs resolution")
ax.set_xlabel("time-grid size N")
ax.set_ylabel("max |δS_h| over endpoint-fixed probes")
ax.legend()
ax.grid(alpha=0.25, which="both")
plt.show()

metrics = {
    "true_variation_residual_finest": float(true_residuals[-1]),
    "wrong_variation_residual_finest": float(wrong_residuals[-1]),
    "true_residual_decreases_with_resolution": bool(np.all(np.diff(true_residuals) < 0.0)),
    "separation_ratio_finest": float(wrong_residuals[-1] / true_residuals[-1]),
}
criteria = {
    "true_variation_residual_finest": "<= 1e-5",
    "wrong_variation_residual_finest": ">= 1e-3",
    "true_residual_decreases_with_resolution": True,
    "separation_ratio_finest": ">= 50",
}
passed = (
    true_residuals[-1] <= 1e-5
    and wrong_residuals[-1] >= 1e-3
    and bool(np.all(np.diff(true_residuals) < 0.0))
    and (wrong_residuals[-1] / true_residuals[-1] >= 50.0)
)
terminal_log("CF14-G3-1", "Primitive stationarity on endpoint-fixed descendant", passed, metrics, criteria, notes="Exact path should approach zero first variation; corrupted path should not.")
append_gate("CF14-G3-1", "Primitive stationarity on endpoint-fixed descendant", "CF14 §4 / primitive stationarity theorem", ["G3", "G6"], passed, metrics, criteria, notes="Mechanical realized descendant of the primitive stationarity theorem.")
```

## Gate CF14-G4-1 — Euler–Lagrange local balance on the same realized mechanical sector

**Plain-language view**

Stationarity is not enough by itself. CF14 also claims the selected path closes a local balance law point by point. This cell attacks that claim directly.

**Canon anchor**
- CF14 §4, Euler–Lagrange equations
- `t h m:euler_lagrange`
- `eq:euler_lagrange`

**Burden type**
- local theorem-bearing balance law

**Exact statement being attacked**
For the realized Lagrangian, the selected path satisfies `d/dt(∂L/∂qdot) - ∂L/∂q = 0`.

**Why this attack is honest**
The notebook computes the pointwise Euler–Lagrange residual on the same exact path and the same corrupted endpoint-fixed control. If the theorem is correct, only the exact path should converge toward zero.

**Pass criteria**
- the exact-path residual decreases with resolution,
- the finest-grid exact residual is below threshold,
- the corrupted-path residual remains finite.

**Fail criteria**
- no convergence on the exact path,
- no separation from the corrupted path.

**This cell cannot prove**
- every possible realized Lagrangian. It audits the local balance theorem on the clean harmonic descendant.

```python
resolutions = np.array([101, 201, 401, 801, 1601], dtype=int)
true_el = []
wrong_el = []

for N in resolutions:
    t = np.linspace(0.0, np.pi, N)
    q_true = np.sin(t)
    q_wrong = np.sin(t) + 0.15 * (t * (np.pi - t) / np.pi**2)

    def el_residual(q: np.ndarray) -> np.ndarray:
        qdot = np.gradient(q, t, edge_order=2)
        qddot = np.gradient(qdot, t, edge_order=2)
        return qddot + q

    r_true = el_residual(q_true)
    r_wrong = el_residual(q_wrong)
    true_el.append(np.max(np.abs(r_true[2:-2])))
    wrong_el.append(np.max(np.abs(r_wrong[2:-2])))

true_el = np.array(true_el)
wrong_el = np.array(wrong_el)

fig, ax = plt.subplots(figsize=(6.8, 3.4))
ax.loglog(resolutions, true_el, marker="o", label="exact path")
ax.loglog(resolutions, wrong_el, marker="o", label="corrupted path")
ax.set_title("CF14-G4 witness: Euler-Lagrange residual vs resolution")
ax.set_xlabel("time-grid size N")
ax.set_ylabel("max |qddot + q|")
ax.legend()
ax.grid(alpha=0.25, which="both")
plt.show()

metrics = {
    "true_el_residual_finest": float(true_el[-1]),
    "wrong_el_residual_finest": float(wrong_el[-1]),
    "true_residual_decreases_with_resolution": bool(np.all(np.diff(true_el) < 0.0)),
    "separation_ratio_finest": float(wrong_el[-1] / true_el[-1]),
}
criteria = {
    "true_el_residual_finest": "<= 2e-6",
    "wrong_el_residual_finest": ">= 1e-2",
    "true_residual_decreases_with_resolution": True,
    "separation_ratio_finest": ">= 1000",
}
passed = (
    true_el[-1] <= 2e-6
    and wrong_el[-1] >= 1e-2
    and bool(np.all(np.diff(true_el) < 0.0))
    and (wrong_el[-1] / true_el[-1] >= 1000.0)
)
terminal_log("CF14-G4-1", "Euler-Lagrange local balance on realized mechanical sector", passed, metrics, criteria, notes="Exact path should satisfy local balance; corrupted path should not.")
append_gate("CF14-G4-1", "Euler-Lagrange local balance on realized mechanical sector", "CF14 §4 / Euler-Lagrange equations", ["G4", "G6"], passed, metrics, criteria, notes="Local path-level non-discharge balance on the harmonic descendant.")
```

## Gate CF14-G7-1 — Thermodynamic Legendre stationarity on a strictly convex model

**Plain-language view**

CF14 says thermodynamic extremisation is not a different rule. On the correct constraint surface, the right thermodynamic potential should become stationary. This cell tests that on an exact convex model.

**Canon anchor**
- CF14 §5.3, Thermodynamic-potential extremisation

**Burden type**
- realized constraint-surface stationarity

**Exact statement being attacked**
At equilibrium, Legendre-transformed thermodynamic potentials have vanishing first variation on the appropriate constraint surface.

**Why this attack is honest**
The notebook uses a strictly convex internal energy `U(S,V)` so the stationary points of `F = U - T S`, `H = U + P V`, and `G = U - T S + P V` are exact and unique. Off-stationary points are checked as negative controls.

**Pass criteria**
- stationary-point gradients vanish to numerical precision,
- Hessians remain positive,
- the off-stationary control keeps a visible gradient.

**Fail criteria**
- stationary points keep nonzero gradients,
- convexity is lost,
- the wrong point looks equally stationary.

**This cell cannot prove**
- every thermodynamic model. It audits the Legendre-stationarity mechanism itself.

```python
a, b, c = 2.0, 3.0, 0.4
T_ext, P_ext = 1.1, -0.6
V_fixed, S_fixed = 0.7, 0.8

A = np.array([[a, c], [c, b]], dtype=float)

def U(S: float, V: float) -> float:
    return 0.5 * a * S**2 + c * S * V + 0.5 * b * V**2

# F(S; T, V_fixed)
S_F = (T_ext - c * V_fixed) / a
grad_F = a * S_F + c * V_fixed - T_ext
hess_F = a

# H(V; S_fixed, P)
V_H = -(c * S_fixed + P_ext) / b
grad_H = c * S_fixed + b * V_H + P_ext
hess_H = b

# G(S, V; T, P)
S_G, V_G = np.linalg.solve(A, np.array([T_ext, -P_ext], dtype=float))
grad_G = np.array([a * S_G + c * V_G - T_ext, c * S_G + b * V_G + P_ext])
hess_G = A.copy()

wrong_point = np.array([S_G + 0.35, V_G - 0.25])
wrong_grad = np.array([
    a * wrong_point[0] + c * wrong_point[1] - T_ext,
    c * wrong_point[0] + b * wrong_point[1] + P_ext,
])

S_grid = np.linspace(S_G - 1.0, S_G + 1.0, 220)
V_grid = np.linspace(V_G - 1.0, V_G + 1.0, 220)
SS, VV = np.meshgrid(S_grid, V_grid, indexing="xy")
GG = U(SS, VV) - T_ext * SS + P_ext * VV

fig, ax = plt.subplots(figsize=(5.6, 4.5))
cs = ax.contour(SS, VV, GG, levels=18)
ax.clabel(cs, inline=True, fontsize=7)
ax.scatter([S_G], [V_G], color="tab:green", label="stationary G point")
ax.scatter([wrong_point[0]], [wrong_point[1]], color="tab:red", label="off-stationary control")
ax.set_title("CF14-G7 witness: Gibbs-potential stationarity")
ax.set_xlabel("S")
ax.set_ylabel("V")
ax.legend(loc="best")
plt.show()

min_hessian_eig = float(min(hess_F, hess_H, np.min(np.linalg.eigvalsh(hess_G))))
max_stationary_grad = float(max(abs(grad_F), abs(grad_H), np.linalg.norm(grad_G, ord=2)))
negative_grad = float(np.linalg.norm(wrong_grad, ord=2))

metrics = {
    "max_stationary_gradient_residual": max_stationary_grad,
    "min_hessian_eigenvalue": min_hessian_eig,
    "negative_control_gradient_norm": negative_grad,
    "S_F": float(S_F),
    "V_H": float(V_H),
    "S_G": float(S_G),
    "V_G": float(V_G),
}
criteria = {
    "max_stationary_gradient_residual": "<= 1e-12",
    "min_hessian_eigenvalue": "> 0",
    "negative_control_gradient_norm": ">= 1e-2",
}
passed = (
    max_stationary_grad <= 1e-12
    and min_hessian_eig > 0.0
    and negative_grad >= 1e-2
)
terminal_log("CF14-G7-1", "Thermodynamic Legendre stationarity on a convex model", passed, metrics, criteria, notes="F, H, and G stationary conditions checked on exact convex data.")
append_gate("CF14-G7-1", "Thermodynamic Legendre stationarity on a convex model", "CF14 §5.3 / thermodynamic-potential extremisation", ["G7"], passed, metrics, criteria, notes="Legendre-surface stationary points separated from an off-surface control.")
```

## Gate CF14-G8-1 — Onsager variational selection for irreversible fluxes

**Plain-language view**

CF14 claims the dissipative limb also admits a variational selection law. This cell checks the exact near-equilibrium form named in the paper.

**Canon anchor**
- CF14 §5.4
- `eq:onsager_dissipation`
- `eq:onsager_constitutive`

**Burden type**
- dissipative realized descendant

**Exact statement being attacked**
Stationarity of `dot(S) - Phi(J)` with respect to fluxes at fixed forces yields the linear constitutive law `X = L J`.

**Why this attack is honest**
The notebook uses a symmetric positive-definite Onsager matrix and a fixed force vector. It attacks the stationary flux with exact gradients, constitutive residuals, and a wrong-flux control.

**Pass criteria**
- the stationary flux gradient vanishes,
- the constitutive residual vanishes,
- the Onsager matrix is positive definite,
- the wrong-flux control keeps a visible gradient.

**Fail criteria**
- nonzero gradient at the claimed stationary flux,
- constitutive mismatch,
- loss of positive definiteness,
- no separation from the wrong-flux control.

**This cell cannot prove**
- nonlinear far-from-equilibrium closures. It audits the exact Onsager descendant written in CF14.

```python
L = np.array([[2.0, 0.3], [0.3, 1.5]], dtype=float)
X = np.array([1.2, -0.7], dtype=float)
J_star = np.linalg.solve(L, X)
J_wrong = np.array([0.2, 0.4], dtype=float)


def onsager_objective(J: np.ndarray) -> float:
    return float(X @ J - 0.5 * J @ L @ J)


def onsager_grad(J: np.ndarray) -> np.ndarray:
    return X - L @ J

xmin, xmax = J_star[0] - 1.4, J_star[0] + 1.4
ymin, ymax = J_star[1] - 1.4, J_star[1] + 1.4
j1 = np.linspace(xmin, xmax, 220)
j2 = np.linspace(ymin, ymax, 220)
J1, J2 = np.meshgrid(j1, j2, indexing="xy")
OBJ = X[0] * J1 + X[1] * J2 - 0.5 * (L[0, 0] * J1**2 + 2.0 * L[0, 1] * J1 * J2 + L[1, 1] * J2**2)

fig, ax = plt.subplots(figsize=(5.6, 4.5))
cs = ax.contour(J1, J2, OBJ, levels=20)
ax.clabel(cs, inline=True, fontsize=7)
ax.scatter([J_star[0]], [J_star[1]], color="tab:green", label="stationary flux")
ax.scatter([J_wrong[0]], [J_wrong[1]], color="tab:red", label="wrong-flux control")
ax.set_title("CF14-G8 witness: Onsager objective in flux space")
ax.set_xlabel("J1")
ax.set_ylabel("J2")
ax.legend(loc="best")
plt.show()

grad_star = onsager_grad(J_star)
grad_wrong = onsager_grad(J_wrong)
constitutive_residual = np.linalg.norm(X - L @ J_star, ord=2)
min_eig_L = float(np.min(np.linalg.eigvalsh(L)))

metrics = {
    "gradient_norm_at_stationary_flux": float(np.linalg.norm(grad_star, ord=2)),
    "constitutive_residual": float(constitutive_residual),
    "negative_control_gradient_norm": float(np.linalg.norm(grad_wrong, ord=2)),
    "onsager_min_eigenvalue": min_eig_L,
    "objective_at_stationary_flux": float(onsager_objective(J_star)),
    "objective_at_wrong_flux": float(onsager_objective(J_wrong)),
}
criteria = {
    "gradient_norm_at_stationary_flux": "<= 1e-12",
    "constitutive_residual": "<= 1e-12",
    "negative_control_gradient_norm": ">= 1e-2",
    "onsager_min_eigenvalue": "> 0",
}
passed = (
    np.linalg.norm(grad_star, ord=2) <= 1e-12
    and constitutive_residual <= 1e-12
    and np.linalg.norm(grad_wrong, ord=2) >= 1e-2
    and min_eig_L > 0.0
)
terminal_log("CF14-G8-1", "Onsager variational selection for irreversible fluxes", passed, metrics, criteria, notes="Exact stationary flux solves X = L J and maximizes dot(S) - Phi.")
append_gate("CF14-G8-1", "Onsager variational selection for irreversible fluxes", "CF14 §5.4 / Onsager dissipative realization", ["G8"], passed, metrics, criteria, notes="Near-equilibrium M-limb variational descendant attacked directly.")
```

## Gate CF14-G9-1 — Generic local field stationarity on a Klein–Gordon descendant

**Plain-language view**

CF14 says the field Euler–Lagrange law is the same root rule expressed on field space. This cell checks that claim on a simple exact periodic mode.

**Canon anchor**
- CF14 §5.5
- `eq:generic_field_action`
- `eq:field_euler_lagrange`

**Burden type**
- generic local field descendant

**Exact statement being attacked**
Applying boundary-fixed variation to a local field action yields the field Euler–Lagrange equation.

**Why this attack is honest**
The notebook uses a periodic Klein–Gordon mode that exactly satisfies the Euler–Lagrange equation and compares it against a nearby wrong-frequency mode. The exact mode should converge toward zero residual under refinement.

**Pass criteria**
- the exact field residual decreases with resolution,
- the finest-grid exact residual is below threshold,
- the wrong-frequency mode remains visibly nonzero.

**Fail criteria**
- no convergence for the exact mode,
- no separation from the wrong mode.

**This cell cannot prove**
- every field theory. It audits the field-space stationarity mechanism on one exact descendant.

```python
resolutions = np.array([32, 64, 128, 256], dtype=int)
k = 2
n_t = 3
m = math.sqrt(n_t**2 - k**2)


def second_periodic(u: np.ndarray, h: float, axis: int) -> np.ndarray:
    return (np.roll(u, -1, axis=axis) - 2.0 * u + np.roll(u, 1, axis=axis)) / (h**2)

true_field_res = []
wrong_field_res = []

for N in resolutions:
    x = np.linspace(0.0, 2.0 * np.pi, N, endpoint=False)
    t = np.linspace(0.0, 2.0 * np.pi, N, endpoint=False)
    dx = x[1] - x[0]
    dt = t[1] - t[0]
    X, T = np.meshgrid(x, t, indexing="xy")
    phi_true = np.sin(k * X) * np.cos(n_t * T)
    phi_wrong = np.sin(k * X) * np.cos(4.0 * T)
    residual_true = second_periodic(phi_true, dt, axis=0) - second_periodic(phi_true, dx, axis=1) + m**2 * phi_true
    residual_wrong = second_periodic(phi_wrong, dt, axis=0) - second_periodic(phi_wrong, dx, axis=1) + m**2 * phi_wrong
    true_field_res.append(np.max(np.abs(residual_true)))
    wrong_field_res.append(np.max(np.abs(residual_wrong)))

true_field_res = np.array(true_field_res)
wrong_field_res = np.array(wrong_field_res)

fig, ax = plt.subplots(figsize=(6.8, 3.4))
ax.loglog(resolutions, true_field_res, marker="o", label="exact KG mode")
ax.loglog(resolutions, wrong_field_res, marker="o", label="wrong-frequency control")
ax.set_title("CF14-G9 witness: field Euler-Lagrange residuals")
ax.set_xlabel("grid size N")
ax.set_ylabel("max PDE residual")
ax.legend()
ax.grid(alpha=0.25, which="both")
plt.show()

metrics = {
    "true_field_residual_finest": float(true_field_res[-1]),
    "wrong_field_residual_finest": float(wrong_field_res[-1]),
    "true_residual_decreases_with_resolution": bool(np.all(np.diff(true_field_res) < 0.0)),
    "separation_ratio_finest": float(wrong_field_res[-1] / true_field_res[-1]),
}
criteria = {
    "true_field_residual_finest": "<= 5e-3",
    "wrong_field_residual_finest": ">= 5",
    "true_residual_decreases_with_resolution": True,
    "separation_ratio_finest": ">= 1000",
}
passed = (
    true_field_res[-1] <= 5e-3
    and wrong_field_res[-1] >= 5.0
    and bool(np.all(np.diff(true_field_res) < 0.0))
    and (wrong_field_res[-1] / true_field_res[-1] >= 1000.0)
)
terminal_log("CF14-G9-1", "Generic local field stationarity on a Klein-Gordon descendant", passed, metrics, criteria, notes="Exact periodic scalar mode should satisfy the discretized field Euler-Lagrange equation.")
append_gate("CF14-G9-1", "Generic local field stationarity on a Klein-Gordon descendant", "CF14 §5.5 / generic local field action", ["G9"], passed, metrics, criteria, notes="Generic field-action descendant attacked via an exact periodic Klein-Gordon mode.")
```

```python
resolutions = np.array([32, 64, 128, 256], dtype=int)
k = 3

true_maxwell_res = []
wrong_maxwell_res = []

for N in resolutions:
    x = np.linspace(0.0, 2.0 * np.pi, N, endpoint=False)
    t = np.linspace(0.0, 2.0 * np.pi, N, endpoint=False)
    dx = x[1] - x[0]
    dt = t[1] - t[0]
    X, T = np.meshgrid(x, t, indexing="xy")
    A_true = np.sin(k * X) * np.cos(k * T)
    A_wrong = np.sin(k * X) * np.cos(4.0 * T)
    residual_true = second_periodic(A_true, dt, axis=0) - second_periodic(A_true, dx, axis=1)
    residual_wrong = second_periodic(A_wrong, dt, axis=0) - second_periodic(A_wrong, dx, axis=1)
    true_maxwell_res.append(np.max(np.abs(residual_true)))
    wrong_maxwell_res.append(np.max(np.abs(residual_wrong)))

true_maxwell_res = np.array(true_maxwell_res)
wrong_maxwell_res = np.array(wrong_maxwell_res)

fig, ax = plt.subplots(figsize=(6.8, 3.4))
ax.loglog(resolutions, true_maxwell_res, marker="o", label="exact vacuum gauge mode")
ax.loglog(resolutions, wrong_maxwell_res, marker="o", label="wrong-frequency control")
ax.set_title("CF14-G9 witness: vacuum Maxwell residuals")
ax.set_xlabel("grid size N")
ax.set_ylabel("max wave residual")
ax.legend()
ax.grid(alpha=0.25, which="both")
plt.show()

metrics = {
    "true_maxwell_residual_finest": float(true_maxwell_res[-1]),
    "wrong_maxwell_residual_finest": float(wrong_maxwell_res[-1]),
    "separation_ratio_finest": float(wrong_maxwell_res[-1] / true_maxwell_res[-1]),
}
criteria = {
    "true_maxwell_residual_finest": "<= 1e-9",
    "wrong_maxwell_residual_finest": ">= 5",
    "separation_ratio_finest": ">= 1e8",
}
passed = (
    true_maxwell_res[-1] <= 1e-9
    and wrong_maxwell_res[-1] >= 5.0
    and (wrong_maxwell_res[-1] / true_maxwell_res[-1] >= 1e8)
)
terminal_log("CF14-G9-2", "Maxwell stationarity on a transverse vacuum mode", passed, metrics, criteria, notes="Exact vacuum mode should satisfy the discretized Maxwell descendant to near machine precision.")
append_gate("CF14-G9-2", "Maxwell stationarity on a transverse vacuum mode", "CF14 §5.6 / Maxwell action realization", ["G9"], passed, metrics, criteria, notes="Gauge descendant attacked in the clean vacuum-wave limit.")
```

## Gate CF14-G10-1 — Stationary-phase witness for the path-integral descendant

**Plain-language view**

CF14 says non-stationary contributions cancel when the phase gets rapid. This cell does not pretend to prove the whole path integral. It tests that core stationary-phase mechanism on an oscillatory integral with one true stationary point.

**Canon anchor**
- CF14 §5.9
- `eq:path_integral`

**Burden type**
- quantum path descendant / asymptotic stationary-phase witness

**Exact statement being attacked**
As the phase scale grows, stationary contributions decay more slowly than off-stationary contributions and therefore dominate.

**Why this attack is honest**
The notebook compares an oscillatory integral centered on the true stationary point `x=0` against a same-width window shifted away from stationarity. The stationary window should show the `λ^{-1/2}` type scaling, while the shifted window should decay faster.

**Pass criteria**
- the stationary-window log-log slope is near `-1/2`,
- the off-stationary slope is more negative,
- the dominance ratio is large on the finest phase scale.

**Fail criteria**
- the stationary contribution does not separate asymptotically,
- both windows decay at the same rate.

**This cell cannot prove**
- the full infinite-dimensional path-integral measure. It is a finite-dimensional stationary-phase witness only.

```python
lambdas = np.array([10, 20, 40, 80, 160, 320], dtype=float)


def osc_int(lam: float, a: float, b: float, N: int = 60000) -> complex:
    x = np.linspace(a, b, N)
    y = np.exp(1j * lam * 0.5 * x**2)
    return np.trapezoid(y, x)

stationary_mag = np.array([abs(osc_int(lam, -1.0, 1.0)) for lam in lambdas])
offstationary_mag = np.array([abs(osc_int(lam, 0.5, 1.5)) for lam in lambdas])
ratio = stationary_mag / offstationary_mag
stationary_slope = float(np.polyfit(np.log(lambdas[-4:]), np.log(stationary_mag[-4:]), 1)[0])
offstationary_slope = float(np.polyfit(np.log(lambdas[-4:]), np.log(offstationary_mag[-4:]), 1)[0])

fig, ax = plt.subplots(figsize=(6.8, 3.4))
ax.loglog(lambdas, stationary_mag, marker="o", label="window containing stationary point")
ax.loglog(lambdas, offstationary_mag, marker="o", label="shifted off-stationary window")
ax.set_title("CF14-G10 witness: stationary-phase separation")
ax.set_xlabel("phase scale λ")
ax.set_ylabel("|oscillatory integral|")
ax.legend()
ax.grid(alpha=0.25, which="both")
plt.show()

metrics = {
    "stationary_window_slope": stationary_slope,
    "off_stationary_window_slope": offstationary_slope,
    "dominance_ratio_finest": float(ratio[-1]),
}
criteria = {
    "stationary_window_slope": "between -0.7 and -0.3",
    "off_stationary_window_slope": "<= -0.8",
    "dominance_ratio_finest": ">= 10",
}
passed = (
    (-0.7 <= stationary_slope <= -0.3)
    and (offstationary_slope <= -0.8)
    and (ratio[-1] >= 10.0)
)
terminal_log("CF14-G10-1", "Stationary-phase witness for the path-integral descendant", passed, metrics, criteria, notes="Finite-dimensional stationary-phase test: stationary window should decay more slowly than the shifted window.")
append_gate("CF14-G10-1", "Stationary-phase witness for the path-integral descendant", "CF14 §5.9 / path integral descendant", ["G10"], passed, metrics, criteria, notes="Finite-dimensional stationary-phase witness, not a full path-integral proof.", partial=True)
```

## Gate CF14-G10-2 — Rayleigh–Ritz stationary-energy descendant on a normalized trial manifold

**Plain-language view**

CF14 explicitly names Rayleigh–Ritz as a state-space descendant. This cell checks the basic claim: stationary energy expectation should line up with eigenstates, and the minimum should land on the ground state.

**Canon anchor**
- CF14 §5.10, Variational quantum state principles

**Burden type**
- quantum state-space descendant

**Exact statement being attacked**
On a normalized trial-state family, stationarity of `⟨ψ|H|ψ⟩` selects eigenstates; the minimum yields the ground-state energy.

**Why this attack is honest**
The notebook uses a symmetric two-level Hamiltonian and a normalized one-angle trial family. The exact stationary angles are known from the eigensystem, and a non-eigenstate angle is used as a negative control.

**Pass criteria**
- derivative residuals at stationary eigen-angles are tiny,
- energies at those angles match the exact eigenvalues,
- the off-stationary control keeps a visible derivative.

**Fail criteria**
- stationary points do not align with eigenstates,
- eigenvalue matching fails,
- the control looks equally stationary.

**This cell cannot prove**
- the full time-dependent variational-principle family. It audits the Rayleigh–Ritz descendant directly named in CF14.

```python
H = np.array([[1.0, 0.2], [0.2, 3.0]], dtype=float)
evals, evecs = np.linalg.eigh(H)

def energy_theta(theta: np.ndarray | float) -> np.ndarray | float:
    c = np.cos(theta)
    s = np.sin(theta)
    return H[0, 0] * c**2 + 2.0 * H[0, 1] * s * c + H[1, 1] * s**2


def denergy_theta(theta: np.ndarray | float) -> np.ndarray | float:
    c = np.cos(theta)
    s = np.sin(theta)
    return 2.0 * (H[1, 1] - H[0, 0]) * s * c + 2.0 * H[0, 1] * (c**2 - s**2)


theta = np.linspace(0.0, 2.0 * np.pi, 2001)
energy = energy_theta(theta)

stationary_thetas = []
stationary_energy_errors = []
stationary_derivatives = []
for idx in range(2):
    v = evecs[:, idx]
    th = math.atan2(v[1], v[0])
    stationary_thetas.append(th)
    stationary_derivatives.append(abs(float(denergy_theta(th))))
    stationary_energy_errors.append(abs(float(energy_theta(th) - evals[idx])))

wrong_theta = np.pi / 4.0
wrong_derivative = abs(float(denergy_theta(wrong_theta)))

fig, ax = plt.subplots(figsize=(6.8, 3.4))
ax.plot(theta, energy)
for th, ev in zip(stationary_thetas, evals):
    th_mod = th % (2.0 * np.pi)
    ax.axvline(th_mod, color="tab:green", alpha=0.6)
    ax.text(th_mod, ev + 0.04, f"eig θ≈{th_mod:.3f}", rotation=90, va="bottom", ha="right", fontsize=8)
ax.axvline(wrong_theta, color="tab:red", alpha=0.6, linestyle="--", label="off-stationary control")
ax.set_title("CF14-G10 witness: Rayleigh-Ritz energy landscape")
ax.set_xlabel("trial-state angle θ")
ax.set_ylabel("⟨ψ(θ)|H|ψ(θ)⟩")
ax.grid(alpha=0.25)
plt.show()

metrics = {
    "stationary_derivative_residual": float(max(stationary_derivatives)),
    "eigenvalue_match_error": float(max(stationary_energy_errors)),
    "negative_control_derivative": wrong_derivative,
    "ground_energy_exact": float(evals[0]),
    "ground_energy_from_landscape": float(np.min(energy)),
}
criteria = {
    "stationary_derivative_residual": "<= 1e-12",
    "eigenvalue_match_error": "<= 1e-12",
    "negative_control_derivative": ">= 1e-2",
}
passed = (
    max(stationary_derivatives) <= 1e-12
    and max(stationary_energy_errors) <= 1e-12
    and wrong_derivative >= 1e-2
)
terminal_log("CF14-G10-2", "Rayleigh-Ritz stationary-energy descendant", passed, metrics, criteria, notes="Stationary trial-state directions should coincide with Hamiltonian eigenvectors.")
append_gate("CF14-G10-2", "Rayleigh-Ritz stationary-energy descendant", "CF14 §5.10 / Rayleigh-Ritz descendant", ["G10"], passed, metrics, criteria, notes="Normalized trial-state manifold audit for the explicit Rayleigh-Ritz descendant.")
```
