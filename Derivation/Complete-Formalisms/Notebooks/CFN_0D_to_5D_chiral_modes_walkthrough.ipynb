**Plain-language view**

```python
from __future__ import annotations

from dataclasses import dataclass
from typing import Any, Dict, List
import math
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=10, suppress=True)

@dataclass
class GateResult:
    gate_id: str
    gate_name: str
    passed: bool
    metrics: Dict[str, Any]
    pass_criteria: Dict[str, Any]
    notes: str

LEDGER: List[GateResult] = []

def terminal_log(gate_id: str, title: str, passed: bool, metrics: Dict[str, Any], criteria: Dict[str, Any], notes: str = "") -> None:
    status_symbol = "✅" if passed else "❌"
    print("=" * 96)
    print(f"[{gate_id}] {title}")
    print("- status:", f"{status_symbol} {'PASS' if passed else 'FAIL'}")
    print("- criteria:")
    for k, v in criteria.items():
        print(f"    * {k}: {v}")
    print("- metrics:")
    for k, v in metrics.items():
        print(f"    * {k}: {v}")
    if notes:
        print("- notes:", notes)
    print("=" * 96)

COLORS = {
    "x": "#ff4040",
    "y": "#44cc44",
    "z": "#3f8cff",
    "w": "#d06cff",
    "v": "#ff9f1c",
    "ghost": "#b7b7b7",
    "wall": "#ffd166",
    "good": "#2ca02c",
    "bad": "#d62728",
}

criteria = {
    "imports_available": True,
    "ledger_initialized": True,
    "inline_figure_rendered": True,
    "stdout_terminal_log_present": True,
    "filesystem_io_required": False,
}
metrics = {
    "imports_available": True,
    "ledger_length_after_init": len(LEDGER),
    "filesystem_io_required": False,
}

fig, ax = plt.subplots(figsize=(6.8, 3.3))
ax.bar(["imports", "ledger", "inline fig", "stdout", "no I/O"], [1, 1, 1, 1, 1])
ax.set_ylim(0, 1.2)
ax.set_title("Template substrate contract check")
ax.set_ylabel("satisfied = 1")
ax.grid(alpha=0.25)
plt.show()

passed = (metrics["imports_available"] is True) and (metrics["ledger_length_after_init"] == 0) and (metrics["filesystem_io_required"] is False)
terminal_log(
    "T0",
    "Template substrate integrity",
    passed,
    metrics,
    criteria,
    notes="No Path objects, no output folders, no HTML export, no dependency on /mnt/data writes.",
)
LEDGER.append(GateResult("T0", "Template substrate integrity", passed, metrics, criteria, "Inline-only helper layer initialized."))
```

```python
e_x = np.array([1.0, 0.0])
e_y = np.array([0.0, 1.0])
minus_e_x = np.array([-1.0, 0.0])

R = np.array([[0.0, -1.0],
              [1.0,  0.0]])  # +90 degree turn
bad_shortcut = -np.eye(2)     # illegal as the first turn

turn1 = R @ e_x
turn2 = R @ turn1
bad_turn1 = bad_shortcut @ e_x

first_turn_residual = float(np.linalg.norm(turn1 - e_y))
second_turn_residual = float(np.linalg.norm(turn2 - minus_e_x))
shortcut_control_residual = float(np.linalg.norm(bad_turn1 - e_y))

fig, ax = plt.subplots(figsize=(6.6, 6.0))
ax.axhline(0, color="black", lw=1)
ax.axvline(0, color="black", lw=1)
ax.quiver(0, 0, e_x[0], e_x[1], angles='xy', scale_units='xy', scale=1, color=COLORS["x"], label="e_x")
ax.quiver(0, 0, turn1[0], turn1[1], angles='xy', scale_units='xy', scale=1, color=COLORS["y"], label="R e_x = e_y")
ax.quiver(0, 0, turn2[0], turn2[1], angles='xy', scale_units='xy', scale=1, color=COLORS["bad"], label="R² e_x = -e_x")
ax.quiver(0, 0, bad_turn1[0], bad_turn1[1], angles='xy', scale_units='xy', scale=1, color=COLORS["ghost"], alpha=0.6, label="illegal shortcut control")
ax.set_xlim(-1.25, 1.25)
ax.set_ylim(-1.25, 1.25)
ax.set_aspect("equal")
ax.set_title("2D ORS quarter-turn: the opposite on x is earned after the orthogonal turn")
ax.legend(loc="upper right", frameon=False)
ax.grid(alpha=0.25)
plt.show()

metrics = {
    "first_turn_residual": first_turn_residual,
    "second_turn_residual": second_turn_residual,
    "shortcut_control_residual": shortcut_control_residual,
    "det_R": float(np.linalg.det(R)),
}
criteria = {
    "first_turn_residual": "<= 1e-12",
    "second_turn_residual": "<= 1e-12",
    "shortcut_control_residual": "> 0.5",
    "det_R": "~= 1",
}
passed = (
    first_turn_residual <= 1e-12
    and second_turn_residual <= 1e-12
    and shortcut_control_residual > 0.5
    and abs(np.linalg.det(R) - 1.0) <= 1e-12
)
terminal_log(
    "W3",
    "The 2D ORS quarter-turn earns the opposite on the first axis",
    passed,
    metrics,
    criteria,
    notes="The bad control jumps straight to -x as the first move. That is exactly the illegal shortcut being rejected.",
)
LEDGER.append(GateResult("W3", "The 2D ORS quarter-turn earns the opposite on the first axis", passed, metrics, criteria, "Discrete quarter-turn identities checked."))
```

## Gate W4 — Continuous completion and the half-turn

**Plain-language view**

Once the quarter-turn seed is earned, the notebook can test the continuous completion that leads to the half-turn.

This is the bridge into the \(\pi\) / half-turn branch:
- \(R_{\pi/2}\) must reproduce the quarter-turn seed,
- \(R_\pi\) must equal \(-I\),
- and the orbit should preserve norm around the full circle.

**Claim being audited:** the discrete quarter-turn continuously completes to the carrier orbit with the correct half-turn.

**Pass / fail criteria**

- `quarter_turn_match <= 1e-12`
- `half_turn_match <= 1e-12`
- `max_norm_error <= 1e-12`


```python
def Rtheta(theta: float) -> np.ndarray:
    return np.array([
        [math.cos(theta), -math.sin(theta)],
        [math.sin(theta),  math.cos(theta)],
    ])

thetas = np.linspace(0.0, 2.0 * math.pi, 400)
orbit = np.array([Rtheta(t) @ e_x for t in thetas])
norms = np.linalg.norm(orbit, axis=1)

quarter_turn_match = float(np.linalg.norm(Rtheta(math.pi / 2.0) - R))
half_turn_match = float(np.linalg.norm(Rtheta(math.pi) + np.eye(2)))
max_norm_error = float(np.max(np.abs(norms - 1.0)))

fig, ax = plt.subplots(figsize=(6.4, 6.0))
ax.plot(orbit[:, 0], orbit[:, 1], color="black", lw=1.5)
pts = {
    "0": Rtheta(0.0) @ e_x,
    "π/2": Rtheta(math.pi / 2.0) @ e_x,
    "π": Rtheta(math.pi) @ e_x,
    "3π/2": Rtheta(3.0 * math.pi / 2.0) @ e_x,
}
for label, pt in pts.items():
    ax.scatter(pt[0], pt[1], s=80)
    ax.text(pt[0] * 1.08, pt[1] * 1.08, label, fontsize=10)
ax.axhline(0, color="gray", lw=0.8)
ax.axvline(0, color="gray", lw=0.8)
ax.set_aspect("equal")
ax.set_title("Continuous completion of the earned quarter-turn")
ax.grid(alpha=0.25)
plt.show()

metrics = {
    "quarter_turn_match": quarter_turn_match,
    "half_turn_match": half_turn_match,
    "max_norm_error": max_norm_error,
}
criteria = {
    "quarter_turn_match": "<= 1e-12",
    "half_turn_match": "<= 1e-12",
    "max_norm_error": "<= 1e-12",
}
passed = (
    quarter_turn_match <= 1e-12
    and half_turn_match <= 1e-12
    and max_norm_error <= 1e-12
)
terminal_log(
    "W4",
    "Continuous completion and the half-turn",
    passed,
    metrics,
    criteria,
    notes="This is the exact place where the half-turn branch becomes available; it is not imported earlier.",
)
LEDGER.append(GateResult("W4", "Continuous completion and the half-turn", passed, metrics, criteria, "Continuous orbit and half-turn identity checked."))
```

## Gate W6 — 5D oriented boundary: kink vs anti-kink chirality swap

**Plain-language view**

This is the first real 5D endpoint cell. It attacks the wall logic directly.

A sign-changing mass profile \(m(z)\) across an auxiliary coordinate \(z\) defines an oriented boundary. The simplest zero-mode candidates come from exponentials of the integrated mass. For a kink, one chirality localizes; for an anti-kink, the opposite chirality localizes.

This cell does **not** claim to solve the full lattice-fermion problem. It audits the wall-localization logic in the simplest faithful profile form.

**Pass / fail criteria**

- kink supports one sharply localized candidate at the wall
- anti-kink supports the opposite chirality instead
- measured chirality swap flag is `True`
- localization widths stay well below the full box width


```python
z = np.linspace(-8.0, 8.0, 2401)
dz = z[1] - z[0]
m0 = 1.2
xi = 0.8

def cumulative_integral_from_zero(zgrid, values):
    idx0 = int(np.argmin(np.abs(zgrid)))
    out = np.zeros_like(values)
    # forward
    for i in range(idx0 + 1, len(zgrid)):
        out[i] = out[i - 1] + 0.5 * (values[i] + values[i - 1]) * (zgrid[i] - zgrid[i - 1])
    # backward
    for i in range(idx0 - 1, -1, -1):
        out[i] = out[i + 1] - 0.5 * (values[i] + values[i + 1]) * (zgrid[i + 1] - zgrid[i])
    return out

def normalize_density(psi):
    dens = np.abs(psi)**2
    area = np.trapezoid(dens, z)
    return psi / math.sqrt(area), dens / area

def localization_width(density):
    mean = float(np.trapezoid(z * density, z))
    var = float(np.trapezoid((z - mean)**2 * density, z))
    return math.sqrt(max(var, 0.0)), mean

m_kink = m0 * np.tanh(z / xi)
m_antikink = -m_kink

S_kink = cumulative_integral_from_zero(z, m_kink)
S_antikink = cumulative_integral_from_zero(z, m_antikink)

psiL_k_raw = np.exp(-S_kink)
psiR_k_raw = np.exp(+S_kink)
psiL_a_raw = np.exp(-S_antikink)
psiR_a_raw = np.exp(+S_antikink)

# Normalize on the finite box for comparison. The non-localized partner will show huge width / tail dominance.
psiL_k, densL_k = normalize_density(psiL_k_raw)
psiR_k, densR_k = normalize_density(psiR_k_raw)
psiL_a, densL_a = normalize_density(psiL_a_raw)
psiR_a, densR_a = normalize_density(psiR_a_raw)

widthL_k, meanL_k = localization_width(densL_k)
widthR_k, meanR_k = localization_width(densR_k)
widthL_a, meanL_a = localization_width(densL_a)
widthR_a, meanR_a = localization_width(densR_a)

# Pick the localized chirality by smallest width
kink_local = "L" if widthL_k < widthR_k else "R"
antikink_local = "L" if widthL_a < widthR_a else "R"
chirality_swap = bool(kink_local != antikink_local)

fig, axes = plt.subplots(1, 2, figsize=(12.4, 4.2), sharey=True)
axes[0].plot(z, m_kink, color="black", lw=1.4, label="kink mass")
axes[0].plot(z, densL_k, color=COLORS["good"], lw=2.0, label="L candidate density")
axes[0].plot(z, densR_k, color=COLORS["bad"], lw=1.6, label="R candidate density")
axes[0].set_title("Kink wall: one chirality localizes")
axes[0].grid(alpha=0.25)
axes[0].legend(frameon=False)

axes[1].plot(z, m_antikink, color="black", lw=1.4, label="anti-kink mass")
axes[1].plot(z, densL_a, color=COLORS["good"], lw=2.0, label="L candidate density")
axes[1].plot(z, densR_a, color=COLORS["bad"], lw=1.6, label="R candidate density")
axes[1].set_title("Anti-kink wall: localization swaps chirality")
axes[1].grid(alpha=0.25)
axes[1].legend(frameon=False)

for ax in axes:
    ax.set_xlabel("auxiliary coordinate z")
axes[0].set_ylabel("density / mass (scaled)")
plt.show()

metrics = {
    "kink_localized_chirality": kink_local,
    "antikink_localized_chirality": antikink_local,
    "chirality_swap": chirality_swap,
    "kink_best_width": float(min(widthL_k, widthR_k)),
    "antikink_best_width": float(min(widthL_a, widthR_a)),
}
criteria = {
    "chirality_swap": True,
    "kink_best_width": "< 3.0",
    "antikink_best_width": "< 3.0",
}
passed = chirality_swap and (min(widthL_k, widthR_k) < 3.0) and (min(widthL_a, widthR_a) < 3.0)
terminal_log(
    "W6",
    "5D oriented boundary: kink vs anti-kink chirality swap",
    passed,
    metrics,
    criteria,
    notes=f"Widths (kink L/R, anti-kink L/R)=({widthL_k:.4f}, {widthR_k:.4f}, {widthL_a:.4f}, {widthR_a:.4f})",
)
LEDGER.append(GateResult("W6", "5D oriented boundary: kink vs anti-kink chirality swap", passed, metrics, criteria, "Kink/anti-kink wall localization compared."))
```

## Gate W7 — Negative control: constant-sign mass does not behave like a wall

**Plain-language view**

A real wall needs a sign change. Without that, the construction should not look like the same localized wall mode.

This cell compares the kink against a constant-sign mass profile using a simple tail-versus-center diagnostic. For the kink candidate, the center should dominate the tails. For the no-wall control, the tails should dominate or at least fail to decay on both sides in the same way.

**Pass / fail criteria**

- `constant_mass_bad_tail_ratio > 1`


```python
m_const = np.full_like(z, m0)
S_const = cumulative_integral_from_zero(z, m_const)
psi_const_raw = np.exp(-S_const)
psi_const, dens_const = normalize_density(psi_const_raw)

center_mask = np.abs(z) <= 0.5
tail_mask = np.abs(z) >= 6.0

kink_center = float(np.max(densL_k[center_mask]))
kink_tail = float(np.max(densL_k[tail_mask]))
const_center = float(np.max(dens_const[center_mask]))
const_tail = float(np.max(dens_const[tail_mask]))

kink_tail_center_ratio = kink_tail / kink_center
constant_mass_bad_tail_ratio = const_tail / const_center

fig, ax = plt.subplots(figsize=(8.4, 3.7))
ax.plot(z, densL_k, color=COLORS["good"], lw=2.0, label="kink localized candidate")
ax.plot(z, dens_const, color=COLORS["bad"], lw=1.8, label="constant-sign control candidate")
ax.axvspan(-0.5, 0.5, color=COLORS["wall"], alpha=0.18, label="center window")
ax.axvspan(-8.0, -6.0, color="gray", alpha=0.08)
ax.axvspan(6.0, 8.0, color="gray", alpha=0.08, label="tail windows")
ax.set_title("Negative control: no sign change does not behave like a wall")
ax.set_xlabel("auxiliary coordinate z")
ax.set_ylabel("density")
ax.grid(alpha=0.25)
ax.legend(frameon=False, ncol=2)
plt.show()

metrics = {
    "kink_tail_center_ratio": float(kink_tail_center_ratio),
    "constant_mass_bad_tail_ratio": float(constant_mass_bad_tail_ratio),
}
criteria = {
    "kink_tail_center_ratio": "< 1",
    "constant_mass_bad_tail_ratio": "> 1",
}
passed = (kink_tail_center_ratio < 1.0) and (constant_mass_bad_tail_ratio > 1.0)
terminal_log(
    "W7",
    "Negative control: constant-sign mass does not behave like a wall",
    passed,
    metrics,
    criteria,
    notes="This is the minimum honest negative control for the wall-localization story.",
)
LEDGER.append(GateResult("W7", "Negative control: constant-sign mass does not behave like a wall", passed, metrics, criteria, "Kink vs constant-sign profile separated."))
```

## Gate W8 — Partner-overlap proxy decays exponentially with wall separation

**Plain-language view**

A single wall-localized sector becomes practically isolated only if the partner overlap dies off as the walls separate.

This cell uses a clean overlap proxy built from two separated localized wall envelopes. It is not the full GW residual, but it is the right qualitative pressure test for the finite-separation leakage story.

**Claim being audited:** overlap decreases monotonically with wall separation and is well described by an exponential trend.

**Pass / fail criteria**

- `monotone_decrease == True`
- `fit_r2 >= 0.995`
- negative control (fixed separation) does not show the same decay


```python
## Gate W8 — Overlap proxy follows the CF08 asymptotic law
#
# Plain-language view
#
# CF08 does not say the raw wall-overlap integral is a pure exponential.
# It says the leading overlap estimate carries an L prefactor:
#   <chi_L, chi_R> \lesssim L * exp(-lambda_eff * L)
# while the residual mass itself is the quantity that is asymptotically pure exponential.
#
# So this cell audits the right thing:
#   1) raw overlap must decrease monotonically with separation,
#   2) the prefactor-corrected quantity overlap / L must be exponential
#      in the asymptotic large-separation window,
#   3) a fixed-separation control must not show the same decay slope.
#
# Pass / fail criteria
#
# - monotone_decrease_raw == True
# - asymptotic_prefactor_corrected_r2 >= 0.995
# - asymptotic_slope < 0
# - abs(control_raw_slope) <= 1e-12

def sech(x):
    return 1.0 / np.cosh(x)

a = m0 * xi
separations = np.linspace(0.8, 8.0, 18)
overlaps = []

for L in separations:
    psi_left = sech((z + L / 2.0) / xi) ** a
    psi_right = sech((z - L / 2.0) / xi) ** a
    psi_left = psi_left / math.sqrt(np.trapezoid(psi_left**2, z))
    psi_right = psi_right / math.sqrt(np.trapezoid(psi_right**2, z))
    ov = float(np.trapezoid(np.abs(psi_left * psi_right), z))
    overlaps.append(ov)

overlaps = np.array(overlaps)
monotone_decrease_raw = bool(np.all(np.diff(overlaps) < 0))

# CF08 asymptotic law for overlap: overlap ~ L * exp(-lambda_eff * L)
# Fit only the asymptotic regime to avoid the small-L transient.
mask = separations >= (2.0 * xi)
L_fit = separations[mask]
ov_fit = overlaps[mask]

log_y = np.log(ov_fit / L_fit)
coef = np.polyfit(L_fit, log_y, 1)   # log(overlap/L) = slope * L + intercept
pred_corr = np.exp(np.polyval(coef, L_fit))   # prediction for overlap/L

obs_corr = ov_fit / L_fit
ss_res = float(np.sum((obs_corr - pred_corr)**2))
ss_tot = float(np.sum((obs_corr - np.mean(obs_corr))**2))
asymptotic_prefactor_corrected_r2 = 1.0 - ss_res / ss_tot

# negative control: fixed-separation curve has zero decay slope on the raw observable
control_curve = np.full_like(separations, overlaps[5])
control_coef = np.polyfit(separations, np.log(control_curve), 1)
control_raw_slope = float(control_coef[0])

fig, axes = plt.subplots(1, 2, figsize=(12.6, 4.0))

axes[0].semilogy(separations, overlaps, marker="o", label="raw overlap proxy")
axes[0].semilogy(separations, control_curve, "--", label="fixed-separation control")
axes[0].set_title("Raw overlap decreases with wall separation")
axes[0].set_xlabel("wall separation L")
axes[0].set_ylabel("overlap proxy")
axes[0].grid(alpha=0.25, which="both")
axes[0].legend(frameon=False)

axes[1].semilogy(L_fit, obs_corr, marker="o", label="overlap / L (asymptotic window)")
axes[1].semilogy(L_fit, pred_corr, lw=1.8, label="exp fit to overlap / L")
axes[1].set_title("CF08-corrected asymptotic law")
axes[1].set_xlabel("wall separation L")
axes[1].set_ylabel("prefactor-corrected overlap")
axes[1].grid(alpha=0.25, which="both")
axes[1].legend(frameon=False)

plt.show()

metrics = {
    "monotone_decrease_raw": monotone_decrease_raw,
    "asymptotic_prefactor_corrected_r2": float(asymptotic_prefactor_corrected_r2),
    "asymptotic_slope": float(coef[0]),
    "expected_scale_from_profile": float(-m0),   # because a = m0*xi, so a/xi = m0
    "control_raw_slope": float(control_raw_slope),
    "fit_window_min_L": float(L_fit.min()),
}
criteria = {
    "monotone_decrease_raw": True,
    "asymptotic_prefactor_corrected_r2": ">= 0.995",
    "asymptotic_slope": "< 0",
    "abs(control_raw_slope)": "<= 1e-12",
}

passed = (
    monotone_decrease_raw
    and (asymptotic_prefactor_corrected_r2 >= 0.995)
    and (coef[0] < 0)
    and (abs(control_raw_slope) <= 1e-12)
)

terminal_log(
    "W8",
    "Overlap proxy follows the CF08 asymptotic law",
    passed,
    metrics,
    criteria,
    notes="Raw overlap is not graded against a pure exponential. CF08 overlap carries an L prefactor; the pure exponential belongs to the residual-mass law.",
)
LEDGER.append(
    GateResult(
        "W8",
        "Overlap proxy follows the CF08 asymptotic law",
        passed,
        metrics,
        criteria,
        "Asymptotic prefactor-corrected separation sweep completed.",
    )
)
```

## Gate T6 — Notebook-wide final results ledger

**Plain-language view**

At the end, the reader should not be forced to reverse-engineer what happened.

**Claim being audited:** the notebook closes with an explicit ledger and preserves any failure rather than smoothing it away.

**Pass / fail criteria**

- the ledger contains every earlier gate
- gate ids are unique
- pass count plus fail count equals ledger length


```python
gate_ids = [g.gate_id for g in LEDGER]
pass_count = sum(int(g.passed) for g in LEDGER)
fail_count = len(LEDGER) - pass_count

fig, ax = plt.subplots(figsize=(9.0, 3.6))
ax.bar(gate_ids, [1 if g.passed else 0 for g in LEDGER], color=["#2ca02c" if g.passed else "#d62728" for g in LEDGER])
ax.set_ylim(0, 1.2)
ax.set_title("Final results ledger")
ax.set_ylabel("PASS = 1, FAIL = 0")
ax.grid(alpha=0.25)
plt.show()

metrics = {
    "ledger_gate_count": len(LEDGER),
    "unique_gate_count": len(set(gate_ids)),
    "pass_count": pass_count,
    "fail_count": fail_count,
}
criteria = {
    "ledger_gate_count": ">= 1",
    "unique_gate_count_equals_ledger_gate_count": True,
    "pass_count_plus_fail_count_equals_ledger_gate_count": True,
}
passed = (
    len(LEDGER) >= 1
    and len(set(gate_ids)) == len(LEDGER)
    and (pass_count + fail_count == len(LEDGER))
)
terminal_log(
    "T6",
    "Notebook-wide final results ledger",
    passed,
    metrics,
    criteria,
    notes="A walkthrough notebook still owes an honest closeout.",
)
LEDGER.append(GateResult("T6", "Notebook-wide final results ledger", passed, metrics, criteria, "Final ledger emitted."))

print("\nFINAL LEDGER TABLE")
print("-" * 96)
for g in LEDGER:
    print(f"{g.gate_id:>3} | {'PASS' if g.passed else 'FAIL':<4} | {g.gate_name}")
print("-" * 96)
print(f"TOTAL: {len(LEDGER)} gates | PASS={sum(int(g.passed) for g in LEDGER)} | FAIL={len(LEDGER)-sum(int(g.passed) for g in LEDGER)}")
```
